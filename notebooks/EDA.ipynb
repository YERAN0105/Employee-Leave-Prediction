# Visualizing Leave Trends

## Leave Distribution per Month

```python
plt.figure(figsize=(10,5))
sns.countplot(x=df['LEAVE MONTH'], palette='coolwarm')
plt.title("Leave Distribution by Month")
plt.xlabel("Month")
plt.ylabel("Number of Leaves")
plt.show()
```

## Leave Frequency per Weekday

```python
plt.figure(figsize=(10,5))
sns.countplot(x=df['LEAVE DAY OF WEEK'], palette='viridis')
plt.title("Leave Frequency by Day of the Week")
plt.xlabel("Day of Week (0=Monday, 6=Sunday)")
plt.ylabel("Number of Leaves")
plt.show()
```

## Leave Type Distribution

```python
plt.figure(figsize=(10,5))
sns.countplot(y=df['TYPE'], palette='Set2')
plt.title("Leave Type Distribution")
plt.xlabel("Count")
plt.ylabel("Leave Type")
plt.show()
```

# Employee Leave Patterns

## Cumulative Leave Count per Employee per Year

```python
df = df.sort_values(by=['NAME', 'LEAVE YEAR', 'LEAVE DATE'])
df['CUMULATIVE LEAVE COUNT'] = df.groupby(['NAME', 'LEAVE YEAR']).cumcount() + 1

plt.figure(figsize=(12,6))
sns.lineplot(data=df, x='LEAVE DATE', y='CUMULATIVE LEAVE COUNT', hue='NAME', legend=False)
plt.title("Cumulative Leave Trend for Employees")
plt.xlabel("Date")
plt.ylabel("Cumulative Leave Count")
plt.show()
```

# Correlation Analysis

```python
# Compute correlation between numerical features
correlation_matrix = df[['LEAVE MONTH', 'LEAVE DAY OF WEEK', 'DAYS SINCE LAST LEAVE', 'DAYS UNTIL NEXT LEAVE']].corr()

# Plot heatmap
plt.figure(figsize=(8,5))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Feature Correlation Heatmap")
plt.show()
```
